```r
# 1. Lista de experimentos y semillas a probar
experimentos <- list(
 list(semilla = 111119, exp = 9194),
 list(semilla = 333337, exp = 9195),
 list(semilla = 444443, exp = 9196),
 list(semilla = 555557, exp = 9197),
 list(semilla = 666667, exp = 9198),
 list(semilla = 100519, exp = 9199),
 list(semilla = 100523, exp = 9200),
 list(semilla = 100537, exp = 9201),
 list(semilla = 100549, exp = 9202),
 list(semilla = 100547, exp = 9203)
)

notebook_base <- "z620_WorkFlow_01_junior_grupoA11.ipynb"

for (item in experimentos) {
  semilla_i <- item$semilla
  exp_i <- item$exp
  cat("\n======================================================\n")
  cat(sprintf("Iniciando Experimento WF%d | Semilla: %d\n", exp_i, semilla_i))
  cat("======================================================\n\n")
  
  # Leemos el notebook original como texto
  lineas <- readLines(notebook_base, warn = FALSE)
  
  # Reemplazamos la semilla y el experimento directamente en el JSON/texto del notebook
  lineas <- gsub("PARAM\\$semilla_primigenia <- [0-9]+", sprintf("PARAM$semilla_primigenia <- %d", semilla_i), lineas)
  lineas <- gsub("PARAM\\$experimento <- [0-9]+", sprintf("PARAM$experimento <- %d", exp_i), lineas)
  
  # Creamos un notebook temporal
  temp_nb <- sprintf("temp_WF%d.ipynb", exp_i)
  writeLines(lineas, temp_nb)
  
  # Ejecutamos el notebook usando jupyter nbconvert nativo
  salida_nb <- sprintf("WF%d_ejecutado.ipynb", exp_i)
  cmd <- sprintf("jupyter nbconvert --to notebook --execute %s --output %s", temp_nb, salida_nb)
  
  res <- system(cmd)
  
  # Eliminamos el archivo temporal
  if (file.exists(temp_nb)) file.remove(temp_nb)
  
  if (res == 0) {
    cat(sprintf("[OK] Experimento WF%d finalizado con éxito.\n", exp_i))
  } else {
    cat(sprintf("[ERROR] Falló la ejecución del experimento WF%d.\n", exp_i))
  }
}
```
